```python
import pandas as pd
import matplotlib.pyplot as plt
```

## 1. Load the raw data

```python
raw = pd.read_csv("data/raw/example_reaction_times.csv")
raw.head()
```

## 2. Process: average reaction time per participant and condition

We save the result to `data/processed/` so later notebooks (or your instructor) can start from it.

```python
summary = (
    raw.groupby(["participant", "condition"], as_index=False)["reaction_time_ms"]
    .mean()
)
summary.to_csv("data/processed/mean_reaction_times.csv", index=False)
summary
```

## 3. Visualize and save the figure

```python
fig, ax = plt.subplots(figsize=(5, 4))
for participant, rows in summary.groupby("participant"):
    ax.plot(rows["condition"], rows["reaction_time_ms"], marker="o", color="grey", alpha=0.6)
ax.set_xlabel("Condition")
ax.set_ylabel("Mean reaction time (ms)")
ax.set_title("Reaction time by condition (one line per participant)")
fig.tight_layout()
fig.savefig("figures/reaction_times.png", dpi=150)
plt.show()
```
